# Stemming

Stemming is the process of reducing a word its word stem that affixes and preflixes or to the root of word known as a lemma. Stemming is important in Netural Language Understanding (NLU) and Natural Language Processing (NLP)

```python
! pip install nltk
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
words = ['eating', 'eats', 'eaten' , 'writing', 'writes', 'programming', 'programs', 'history', 'finally', 'finalized']
```

```python
for word in words:
  print(word+"----->"+stemming.stem(word))
```

```python
stemming.stem('congratulation')
```

```python
stemming.stem('sitting')
```

## RegexpStemmer Class

NLTK has RegexpStemmer Class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a singel regular expression and remove any prefix or suffix that match the expression. Let us see an example.

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingplaying')
```

```python
for word in words:
  print(word+"----->"+reg_stemmer.stem(word))
```

# Wordnet Lemmatization

Lemmatization technique is like stemming. The output we will get after lemmatization is called "Lemma", which is a root stem, the output of stemming. After lemmatization, we will br getting a valid worf that means the same thing.


NLTK provide WordNetLemmatization class which is a thin wrapper around the wordnet corpus. This class uses morpy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
"""
Part of speach

Noun-n
verb-v
adjective-a
adverb-r

"""

for word in words:
  print(word+"----->"+lemmatizer.lemmatize(word, pos='v'))
```

```python
lemmatizer.lemmatize('Zainul')
```
